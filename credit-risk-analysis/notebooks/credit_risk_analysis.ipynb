# Statistical Credit Risk Analysis and Loan Default Prediction

## 1. Project Goal

The objective of this project is to investigate the statistical determinants of loan default and estimate the probability of default using loan applicant data. The analysis combines statistical inference, exploratory data analysis and predictive modeling techniques to gain insight into borrower risk profiles.

## 2. Credit Risk Background

Credit risk arises from the possibility that a borrower fails to repay a loan according to the agreed terms. Since such events may lead to substantial financial losses, assessing borrower risk is a fundamental component of the lending process.

One of the key quantities in credit risk management is the Probability of Default (PD), which quantifies the likelihood that a borrower will default over a specified time horizon. As such, PD serves as a fundamental measure of borrower creditworthiness and constitutes a central component of credit risk assessment.

From a statistical perspective, credit risk assessment can be formulated as a binary classification problem, where borrowers are categorized as either defaulting or non-defaulting. Consequently, understanding the statistical properties of borrower characteristics and identifying the factors associated with default behavior are essential steps before constructing predictive models.
